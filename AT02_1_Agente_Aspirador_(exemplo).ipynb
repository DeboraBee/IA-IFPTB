# 🤖 Inteligência Artificial — Agente Aspirador

## Atividade demonstrativa — Agentes Inteligentes

### Objetivos

Ao final desta atividade, você deverá ser capaz de:

- representar percepções;
- implementar regras simples de decisão;

---

## 1. O problema

Imagine um pequeno robô aspirador que circula entre dois ambientes:

```text
┌───────────┬───────────┐
│     A     │     B     │
│           │           │
└───────────┴───────────┘
```

Cada ambiente pode estar:

- `Limpo`
- `Sujo`

O agente pode:

- `Aspirar`
- `Esquerda`
- `Direita`
- `Parar`

A cada instante, o agente:

1. percebe onde está;
2. verifica se o ambiente está sujo;
3. escolhe uma ação;
4. executa a ação;
5. recebe uma nova percepção.

---

## 2. Criando o ambiente

Execute a célula abaixo.

````python
ambiente = {
    "A": "Sujo",
    "B": "Sujo"
}

posicao = "A"

print("Ambiente inicial:")
print(ambiente)
print("Posição inicial:", posicao)
````

---

## 3. Criando o agente

Vamos implementar uma função que representa o comportamento do agente.

````python
def agente(posicao, estado):
    if estado == "Sujo":
        return "Aspirar"

    if posicao == "A":
        return "Direita"

    return "Esquerda"
````

Observe que o agente utiliza a percepção para escolher uma ação.


---

## 4. Testando o agente

````python
print(agente("A", "Sujo"))
print(agente("A", "Limpo"))
print(agente("B", "Limpo"))
````

**Observe os resultados:**

O agente não executa sempre a mesma ação.

Sua ação depende da percepção recebida.

---

## 5. Simulando o ambiente

Agora vamos executar o agente durante várias etapas.

````python
ambiente = {
    "A": "Sujo",
    "B": "Sujo"
}

posicao = "A"

pontuacao = 0

for passo in range(10):

    estado = ambiente[posicao]

    acao = agente(posicao, estado)

    print(f"Passo {passo + 1}")
    print(f"Posição: {posicao}")
    print(f"Percepção: {estado}")
    print(f"Ação: {acao}")

    if acao == "Aspirar":
        ambiente[posicao] = "Limpo"
        pontuacao += 1

    elif acao == "Direita":
        posicao = "B"

    elif acao == "Esquerda":
        posicao = "A"

    print("Ambiente:", ambiente)
    print()
````

---

## 6. Medida de desempenho

Um agente racional precisa de uma forma de avaliar seu desempenho.

Neste exemplo anterior, utilizamos uma medida bastante simples:

> **+1 ponto para cada ambiente que o agente limpar.**

Vamos verificar a pontuação.

````python
print("Pontuação final:", pontuacao)
````

### Pergunta para reflexão

A pontuação utilizada é uma boa medida de desempenho?

Pense:

- O agente gastou energia?
- O agente perdeu tempo?
- O agente poderia ficar alternando entre os ambientes?
- O ambiente poderia ficar sujo novamente?

## 7. Melhorando a simulação

Vamos criar uma função para executar uma simulação completa.

````python
def executar_simulacao(ambiente_inicial, posicao_inicial, passos=10):

    ambiente = ambiente_inicial.copy()
    posicao = posicao_inicial
    pontuacao = 0

    for passo in range(passos):

        estado = ambiente[posicao]
        acao = agente(posicao, estado)

        if acao == "Aspirar":
            ambiente[posicao] = "Limpo"
            pontuacao += 1

        elif acao == "Direita":
            posicao = "B"

        elif acao == "Esquerda":
            posicao = "A"

    return ambiente, posicao, pontuacao
````

Teste:


````python
resultado = executar_simulacao(
    {"A": "Sujo", "B": "Sujo"},
    "A"
)

print("Ambiente final:", resultado[0])
print("Posição final:", resultado[1])
print("Pontuação:", resultado[2])
````

# 10. Desafio rápido

Copie o notebook e modifique a função `agente()` para que o robô:

1. aspire quando encontrar sujeira;
2. permaneça no ambiente se ele estiver limpo;
3. só se mova para o outro ambiente quando necessário.

Depois compare o comportamento com a versão original.

````python
# Resposta ao desafio 10

def criar_agente_com_memoria():
    """
    Cria um agente baseado em modelo (model-based agent):
    ele guarda na memória o que já percebeu sobre cada ambiente.
    """
    memoria = {"A": None, "B": None}  # None = ainda não percebido

    def agente(posicao, estado):
        # Atualiza a memória com a percepção atual
        memoria[posicao] = estado

        # Regra 1: se está sujo aqui, aspira
        if estado == "Sujo":
            return "Aspirar"

        # Regra 2: já sei o estado dos dois ambientes?
        outra_posicao = "B" if posicao == "A" else "A"
        estado_outra = memoria[outra_posicao]

        # Se ainda não conheço o outro ambiente, preciso ir checar
        if estado_outra is None:
            return "Direita" if posicao == "A" else "Esquerda"

        # Se o outro ambiente está sujo, vou até lá
        if estado_outra == "Sujo":
            return "Direita" if posicao == "A" else "Esquerda"

        # Se os dois ambientes já são conhecidos e estão limpos, paro
        return "Parar"

    return agente
````

````python
# Resposta ao desafio 10
# Simulação

def executar_simulacao_v2(ambiente_inicial, posicao_inicial, passos=10):
    ambiente = ambiente_inicial.copy()
    posicao = posicao_inicial
    pontuacao = 0
    agente = criar_agente_com_memoria()

    for passo in range(passos):
        estado = ambiente[posicao]
        acao = agente(posicao, estado)

        print(f"Passo {passo+1} | Posição: {posicao} | Percepção: {estado} | Ação: {acao}")

        if acao == "Aspirar":
            ambiente[posicao] = "Limpo"
            pontuacao += 1
        elif acao == "Direita":
            posicao = "B"
        elif acao == "Esquerda":
            posicao = "A"
        elif acao == "Parar":
            print("Agente concluiu a tarefa e parou.")
            break

    return ambiente, posicao, pontuacao, passo + 1
````

````python
# Resposta desafio 10

resultado = executar_simulacao_v2({"A": "Sujo", "B": "Sujo"}, "A")
print("\nAmbiente final:", resultado[0])
print("Posição final:", resultado[1])
print("Pontuação:", resultado[2])
print("Passos usados:", resultado[3])
````

# 11. Reflexão final

Responda:

### Questão 1
Quais são os sensores do agente nesse exemplo?

### Resposta
O sensor continua sendo a leitura de ambiente[posicao] (posição + estado de sujeira do local atual). A diferença é que agora essa percepção é armazenada em memoria, então o agente também "enxerga" indiretamente, ao longo do tempo, o estado do ambiente inteiro — não só o local em que está no instante presente.

### Questão 2
Quais são seus atuadores?

### Resposta
As mesmas três ações de antes — Aspirar, Direita, Esquerda — mais uma nova, Parar, que agora é efetivamente usada como atuador de encerramento quando não há mais trabalho a fazer.

### Questão 3
Qual é a percepção recebida pelo agente?

### Resposta
Igual à versão anterior no formato (posicao, estado), mas seu papel mudou: cada percepção não é só usada para decidir a ação imediata, ela também atualiza um modelo interno do mundo (memoria). É a diferença entre um agente reflexo simples e um agente baseado em modelo.


### Questão 4
Qual é sua medida de desempenho?

### Resposta
Ainda é pontuacao (quantidade de sujeira removida), mas agora faz sentido considerar também uma segunda métrica implícita no código: número de passos usados (passo + 1). Isso captura eficiência, algo que a métrica original ignorava completamente.

### Questão 5
O agente é racional? Justifique.

### Resposta
Agora, ele escolhe sempre a melhor ação — inclusive a de parar quando não há mais ganho possível de desempenho. Ele não desperdiça ações depois de atingir o objetivo, o que era exatamente a falha de racionalidade apontada na versão anterior. Continua limitado pelo que consegue perceber (só conhece o que já visitou), mas dentro dessas limitações ele age de forma consistente e eficiente.

### Questão 6
Quais limitações existem nesse agente?

### Resposta

Só funciona bem em ambientes estáticos: se um ambiente pudesse voltar a ficar sujo depois de limpo, a memória ficaria desatualizada e o agente pararia erroneamente.
Não escala: a lógica de checar "conheço o outro ambiente?" foi escrita à mão para exatamente 2 posições; com N ambientes precisaria de uma estrutura de busca/planejamento mais genérica.
Ainda é reativo por regras fixas, não por otimização — ele não calcula o caminho mais curto entre múltiplos ambientes sujos, apenas segue uma heurística simples de ir para o vizinho.
Memória nunca é limpa: se o ambiente mudasse dinamicamente durante a execução (fora do controle do agente), não haveria mecanismo para invalidar percepções antigas.
