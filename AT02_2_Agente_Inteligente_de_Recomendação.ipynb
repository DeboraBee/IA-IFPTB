# 4. Desenvolvendo o agente

Complemente a implementação da função abaixo:

````python
def recomendar_filme(usuario, filmes):
    # resposta função
    compativeis = [
        f for f in filmes
        if f["duracao"] <= usuario["duracao_maxima"]
        and f["classificacao"] <= usuario["classificacao_maxima"]
    ]

    if not compativeis:
        return None

    do_genero_preferido = [f for f in compativeis if f["genero"] == usuario["genero"]]
    candidatos = do_genero_preferido if do_genero_preferido else compativeis

    return max(candidatos, key=lambda f: f["avaliacao"])
````

A função deverá retornar o filme escolhido pelo agente.

# 8. Criando uma função de pontuação

Agora transforme seu agente em um sistema de decisão mais sofisticado.

Crie uma função que atribua uma pontuação aos filmes.

Por exemplo:

```text
+5 pontos → gênero preferido
+3 pontos → avaliação ≥ 8.5
+2 pontos → duração ≤ 120 minutos
+1 ponto  → duração ≤ 150 minutos
```

Você pode criar sua própria estratégia.

A decisão final deverá considerar a **pontuação total**.

````python
def calcular_pontuacao(filme, usuario):
    """
    +5 pontos -> gênero igual ao preferido pelo usuário
    +3 pontos -> avaliação >= 8.5
    +2 pontos -> duração <= 120 minutos
    +1 ponto  -> duração <= 150 minutos (e > 120)
    """
    pontos = 0
    if filme["genero"] == usuario["genero"]:
        pontos += 5
    if filme["avaliacao"] >= 8.5:
        pontos += 3
    if filme["duracao"] <= 120:
        pontos += 2
    elif filme["duracao"] <= 150:
        pontos += 1
    return pontos

def recomendar_filme_pontuacao(usuario, filmes):
    compativeis = [
        f for f in filmes
        if f["duracao"] <= usuario["duracao_maxima"]
        and f["classificacao"] <= usuario["classificacao_maxima"]
        and f["avaliacao"] >= usuario.get("avaliacao_minima", 0)
    ]
    if not compativeis:
        return None, []
    pontuados = [(f, calcular_pontuacao(f, usuario)) for f in compativeis]
    pontuados.sort(key=lambda x: x[1], reverse=True)
    return pontuados[0][0], pontuados
````

````python
# Teste
escolhido, ranking = recomendar_filme_pontuacao(usuario, filmes)
print("Filme recomendado (por pontuação):", escolhido["titulo"])
print("\nRanking de pontuação dos filmes compatíveis:")
for f, p in ranking:
    print(f" - {f['titulo']:<18} pontuação: {p}  (avaliação: {f['avaliacao']}, duração: {f['duracao']})")
````

A Estratégia 2 (pontuação) representa melhor um agente racional. Isso é coerente com a definição de racionalidade trabalhada nas questões anteriores: um agente racional escolhe a ação (aqui, a recomendação) que maximiza a medida de desempenho definida, considerando todas as percepções disponíveis — e a Estratégia 2 usa mais dessas percepções na decisão do que a Estratégia 1.

Nos testes realizados com a base de dados original, as duas estratégias coincidiram no resultado (Matrix foi escolhido em ambas), pois esse filme já era superior nos dois critérios ao mesmo tempo. Isso não significa que sejam equivalentes — apenas que, para este conjunto específico de dados, não houve conflito entre "maior avaliação" e "maior pontuação combinada".

# 12. Desafio ⭐

Imagine que o agente tenha acesso ao histórico do usuário:

````python
historico = [
    "Matrix",
    "Interestelar",
    "Top Gun"
]
````

Modifique o agente para considerar o histórico.

Por exemplo:

> Filmes de gêneros semelhantes aos já assistidos podem receber maior pontuação.

### Pergunta

O agente agora está utilizando apenas a percepção atual ou também possui alguma forma de **memória/estado interno**?

Explique.

### Resposta

O agente passou a utilizar uma forma de estado interno/memória. Na versão da seção 8, a pontuação de um filme dependia apenas dos atributos daquele filme comparados às preferências do usuário — uma percepção isolada, sem relação com decisões ou dados anteriores. Na versão com histórico, a função generos_do_historico constrói uma representação a partir de escolhas passadas do usuário (historico), e essa representação é usada para modificar a pontuação de filmes atuais. Isso corresponde ao conceito de agente baseado em modelo: a decisão não depende só da percepção instantânea (os dados do filme sendo avaliado), mas também de uma informação acumulada sobre o que já aconteceu antes (os filmes já assistidos), que persiste e influencia decisões futuras.

````python
# Considerando o histórico

def generos_do_historico(historico, base_filmes):
    generos = set()
    for titulo in historico:
        for f in base_filmes:
            if f["titulo"] == titulo:
                generos.add(f["genero"])
    return generos

def calcular_pontuacao_com_historico(filme, usuario, generos_historico):
    """
    Mesma pontuação da seção 8, acrescida de:
    +2 pontos -> gênero do filme já aparece no histórico do usuário
    """
    pontos = calcular_pontuacao(filme, usuario)
    if filme["genero"] in generos_historico:
        pontos += 2
    return pontos

def recomendar_filme_com_historico(usuario, filmes, historico):
    compativeis = [
        f for f in filmes
        if f["duracao"] <= usuario["duracao_maxima"]
        and f["classificacao"] <= usuario["classificacao_maxima"]
        and f["avaliacao"] >= usuario.get("avaliacao_minima", 0)
    ]
    if not compativeis:
        return None, []
    generos_hist = generos_do_historico(historico, filmes)
    pontuados = [(f, calcular_pontuacao_com_historico(f, usuario, generos_hist)) for f in compativeis]
    pontuados.sort(key=lambda x: x[1], reverse=True)
    return pontuados[0][0], pontuados
````

````python
# Teste
print("Histórico:", historico)
print("Gêneros identificados no histórico:", generos_do_historico(historico, filmes_ampliado))

escolhido_hist, ranking_hist = recomendar_filme_com_historico(usuario_a, filmes_ampliado, historico)
print("\nFilme recomendado (com histórico):", escolhido_hist["titulo"])
print("Ranking com histórico:")
for f, p in ranking_hist:
    print(f" - {f['titulo']:<18} pontuação: {p}")
````

````python
# Teste de comparação - sem histórico
escolhido_sem_hist, ranking_sem_hist = recomendar_filme_pontuacao(usuario_a, filmes_ampliado)
print("\nFilme recomendado (sem histórico):", escolhido_sem_hist["titulo"])
print("Ranking sem histórico:")
for f, p in ranking_sem_hist:
    print(f" - {f['titulo']:<18} pontuação: {p}")
````

# 14. Reflexão final

Responda:

### 1.
O sistema pode ser considerado um agente inteligente? Por quê?

### Resposta
Sim, mas com uma ressalva importante. O sistema atende à estrutura básica de um agente: percebe o ambiente (dados dos filmes, preferências do usuário e, na versão da seção 12, também o histórico), processa essa percepção segundo uma lógica de decisão, e executa uma ação (recomendar um filme). No entanto, diferente do agente aspirador de pó, que operava em um ciclo contínuo de percepção → decisão → ação → nova percepção dentro de uma simulação com múltiplos passos, o sistema de recomendação percebe uma única vez e age uma única vez, encerrando a execução em seguida. Não há um laço em que a ação do agente (a recomendação feita) gere uma nova percepção do ambiente já modificado por essa ação, como acontecia quando o aspirador limpava um ambiente e, no passo seguinte, percebia esse mesmo ambiente já como "Limpo". Ainda assim, considerando a versão da seção 12, o sistema pode ser classificado como um agente baseado em modelo, já que incorpora historico como uma forma de estado interno que influencia a decisão, mesmo sem repetir o ciclo automaticamente.

### 2.
O agente sempre escolhe o "melhor" filme?

### Resposta
Não necessariamente. O agente escolhe o filme com maior pontuação segundo `calcular_pontuacao_com_historico`, que é uma função definida com critérios e pesos fixos (gênero, avaliação, duração, semelhança com o histórico). Além disso, como o agente não repete o ciclo de perceber-decidir-agir, ele também não tem a chance de reavaliar ou corrigir sua escolha com base em uma reação do usuário à recomendação — ele decide uma única vez, com as informações que tinha naquele momento, sem retroalimentação.

### 3.
O que significa "melhor" nesse contexto?

### Resposta
Significa o filme com maior valor segundo a medida de desempenho implementada em `calcular_pontuacao_com_historico`: a soma dos pontos por gênero preferido, avaliação, duração e correspondência com os gêneros presentes no `historico`. Não é um julgamento absoluto de qualidade do filme, mas o resultado de uma métrica específica, calculada uma única vez a partir de um `historico` estático fornecido como entrada.

### 4.
Quem define a medida de desempenho do agente?

### Resposta
o usuário define as preferências e restrições que alimentam o cálculo (o quê), mas quem define a medida de desempenho propriamente dita — a fórmula, os critérios escolhidos e o peso de cada um (o como) — é quem projetou o agente. Um sistema mais alinhado à ideia de agente verdadeiramente racional em relação às preferências do usuário precisaria permitir que o próprio usuário calibrasse esses pesos, e não apenas os parâmetros de filtro.

### 5.
O que aconteceria se as informações sobre o usuário fossem incompletas?

### Resposta
Se `historico` fosse uma lista vazia, `generos_do_historico` retornaria um conjunto vazio, e nenhum filme receberia o bônus de +2 pontos — a função se comportaria exatamente como a versão da seção 8, sem quebrar a execução. Já a ausência de chaves obrigatórias em usuario (como `genero`, `duracao_maxima` ou `classificacao_maxima`) geraria um erro (`KeyError`), pois essas chaves são acessadas diretamente no código, sem uso de `.get()` com valor padrão como ocorre com `avaliacao_minima`.

### 6.
Como o agente poderia aprender com as escolhas anteriores do usuário?

### Resposta
Na versão da seção 12, isso ocorre de forma limitada: o agente lê o `historico` (uma lista fixa, definida manualmente antes da execução) e usa os gêneros presentes nele para dar bônus de pontuação a filmes semelhantes. Isso não é um aprendizado automático nem contínuo — é uma regra fixa aplicada sobre um dado histórico estático. Para que o agente realmente "aprendesse" ao longo do tempo, seria necessário que o próprio `historico` fosse atualizado automaticamente a cada nova recomendação aceita pelo usuário, criando o ciclo de retroalimentação (ação influenciando a próxima percepção) que esteve ausente neste exercício, mas que era justamente o elemento central da simulação do aspirador de pó.

O usuário só fornece parâmetros de entrada (gênero, duração máxima, avaliação mínima); os pesos e critérios que realmente decidem qual filme "vence" (+5 para gênero, +3 para avaliação, +2 para duração curta, +2 para histórico) são fixos no código, iguais para qualquer usuário. O `historico` usado como estado interno é um dado estático fornecido antes da execução, não algo que o próprio agente atualiza automaticamente a partir das recomendações que ele mesmo fez. Ou seja, o sistema simula uma memória, mas não fecha o laço de retroalimentação que caracterizava a inteligência do agente na atividade anterior.

Por isso, embora a comparação entre a Estratégia 1 (maior avaliação) e a Estratégia 2 (pontuação combinada) mostre que incorporar mais critérios na medida de desempenho torna a decisão mais completa, isso não torna o agente mais alinhado às preferências reais do usuário — apenas mais completo em relação aos critérios que o projetista decidiu incluir. Um sistema de recomendação verdadeiramente racional em relação ao usuário exigiria não apenas mais critérios, mas a capacidade de ajustar os pesos desses critérios a partir do próprio usuário e de repetir o ciclo de decisão incorporando o resultado das recomendações anteriores — nenhuma das duas coisas está presente na implementação feita nesta atividade.
